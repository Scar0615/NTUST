```python
# Example 1: point plot(折線圖)
# 這裡畫了一個y = x^2的圖

from matplotlib import pyplot

x = [i for i in range(0,51,10)]
y = [i**2 for i in x]

pyplot.plot(x,y, marker='x') # marker是指定每個點的圖案(只有o,x可以)
pyplot.show()
```

```python
# Example 2: point plots with description

import numpy as np
import matplotlib.pyplot as plt
import random

random.seed()

x = np.arange(0, 4*np.pi, 0.1)
y1 = np.sin(x)
y2 = np.sin(2*x)/4+1
y3 = [random.uniform(-1, 1)*1.2 for _ in x] # for _ in x其實和for i in x是一樣的 但是我們這邊不需要用到iterator 所以用底線

fig, ax = plt.subplots(layout = 'constrained')

# lw -> line width
ax.plot(x, y1, color = 'green', lw = 0.5, marker = 'x', ms = 4, label = '$f_1(x)$')
ax.plot(x, y2, color = 'k', lw = 2, label = '$f_2(x)$')
ax.plot(x, y3, color = 'r', lw = 1, label = '$f_3(x)$')

ax.set_ylabel('y(x)')
ax.set_xlabel('x')
ax.set_title('Example function y(x)')
ax.legend(loc = 'upper left', ncols = 2)
ax.set_ylim(-1.5, 2) # 設定y軸的上下限

plt.show()
```

```python
# Example 3: CDF
import matplotlib.pyplot as plt
#import pandas as pd
import numpy as np
import random
from statsmodels.distributions.empirical_distribution import ECDF

fig, ax = plt.subplots(layout='constrained')

# raw data
x = [random.uniform(0,100) for _ in range(100)]

# CDF生成方法一：manually computed CDF
cdf_x = np.sort(x)
cdf_y = (np.arange(len(x))+1)/(len(x))
ax.plot(cdf_x, cdf_y, marker='x', label='manual')

# CDF生成方法二：one-line solution in Matplotlib
# 哎...我用不了ECDF()，我還是用第一個方法就好。
# ax.ecdf(x, label='ECDF()')

# 繪圖
ax.set_ylim(0, 1.05)
ax.legend()
ax.set_ylabel('Probability')
ax.set_xlabel('data value')
ax.set_title('Example CDF')
plt.show()
```

```python
# Example 4: average and confidence interval
# 95%信賴區間指的是在下一次實驗時 下一次的結果有95%的機率會在這個區間裡

import matplotlib.pyplot as plt
import numpy as np
import random

dataset = []
cases = 3
for _ in range(cases):
    x = [random.uniform(20,100) for _ in range(2)]
    dataset.append(x)
print(dataset)

avg = []
con = []
for data in dataset:
    # computing the average value
    avg.append(np.mean(data))
    # computing the 95% confidence interval; see https://www.scribbr.com/statistics/confidence-interval/
    con.append(1.96 * np.std(data, ddof=1) / np.sqrt(len(data)))

nx = ['case 1', 'case 2', 'case 3'] 
plt.errorbar(nx, avg, yerr=con,
             lw=0.7, elinewidth=0.7, marker='s', capsize=2, markersize=5,
             color='g', label='example')
plt.ylim(-50, 150)
plt.show()
```
